```python
class EditorTexto:
    def __init__(self):
        self.texto = ""
        self.pilha = []

    def escrever(self, texto):
        # Salva o estado atual antes da alteração
        self.pilha.append(self.texto)

        self.texto += texto
        print(f"Texto atual: {self.texto}")

    def apagar(self, quantidade):
        self.pilha.append(self.texto)

        self.texto = self.texto[:-quantidade]
        print(f"Texto atual: {self.texto}")

    def desfazer(self):
        if len(self.pilha) > 0:
            self.texto = self.pilha.pop()
            print(f"Desfeito! Texto atual: {self.texto}")
        else:
            print("Não há ações para desfazer.")

    def mostrar_texto(self):
        print(f"Texto: {self.texto}")

editor = EditorTexto()

editor.escrever("Olá ")
editor.escrever("mundo!")
editor.mostrar_texto()

editor.desfazer()
editor.mostrar_texto()

editor.desfazer()
editor.mostrar_texto()
```

```python
class SistemaImpressao:

    def __init__(self):
        self.fila = []

    def adicionar(self, documento):
        self.fila.append(documento)
        print(f"Documento {documento} está sendo adicionado à fila")

    def imprimir(self):
        if len(self.fila) > 0:
            documento = self.fila.pop(0)
            print(f"Imprimindo {documento}")
        else:
            print("Não tem o que imprimir")

    def mostrar_fila(self):
        if len(self.fila) > 0:
            print("\nDocumentos na fila:")
            for documento in self.fila:
                print(f"- {documento}")
        else:
            print("\nA fila está vazia.")


Fila = SistemaImpressao()

Fila.adicionar("Trabalho.pdf")
Fila.adicionar("Algo.pdf")
Fila.adicionar("Flamengo.pdf")

Fila.mostrar_fila()

Fila.imprimir()
Fila.imprimir()

Fila.mostrar_fila()
```
